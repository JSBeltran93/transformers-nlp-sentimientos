Este notebook usa un modelo Transformer preentrenado (DistilBERT afinado para análisis de sentimientos) a través de la librería `transformers` de Hugging Face, para clasificar el sentimiento de un conjunto de oraciones de prueba.

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis",
                         model="distilbert-base-uncased-finetuned-sst-2-english")
```

## Oraciones de prueba

Se eligieron 5 oraciones en inglés (idioma en el que el modelo fue afinado): dos claramente positivas, dos claramente negativas, y una neutra/ambigua a propósito, para observar cómo se comporta el modelo en un caso límite donde no hay una carga emocional clara.

```python
oraciones = [
    "I love how this model understands context so well.",
    "This movie was a complete waste of my time.",
    "The weather today is okay, nothing special.",
    "Transformers have completely changed the field of NLP.",
    "I am really disappointed with the customer service.",
]

resultados = clasificador(oraciones)
for oracion, resultado in zip(oraciones, resultados):
    print(f"{oracion!r} -> {resultado['label']} ({resultado['score']:.4f})")
```

## Observación

Las cuatro oraciones con una carga emocional clara se clasificaron correctamente con una confianza casi perfecta. La oración neutra ("The weather today is okay, nothing special") fue clasificada como NEGATIVE con 99.52% de confianza — un resultado esperable, ya que el modelo fue entrenado sobre un dataset binario (SST-2) que no contempla una clase "neutral", por lo que ante un caso ambiguo el modelo igual produce una respuesta segura hacia uno de los dos extremos.
